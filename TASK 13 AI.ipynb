```python
def convertObjToInt(cols):
    for i in cols:
        data = pd.DataFrame({i:df[i].unique()})
        data_LE = LabelEncoder()
        data_LE.fit(np.ravel(data))    # data_LE.fit(data)
        df[i] = data_LE.transform(df[i])
        
columns = ['School DBN']
convertObjToInt(columns)
```

```python
df.drop("Released", axis=1, inplace=True)
```

```python
df.dtypes
```

```python
import pandas as pd
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import pickle
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

```python
X = df.drop('Present', axis=1)   # Dependant columns / Input columns
y = df['Present']                # Resultant columns / Output column
```

```python
from sklearn.model_selection import train_test_split
train_X, test_X, train_y, test_y = train_test_split(X,y, test_size=0.3, shuffle=False) # 80:20, 70:30, 60:40
```

```python
model = KNeighborsRegressor(n_neighbors=2)
model.fit(X, y)

```

```python
y_pred = model.predict(X)
y_pred
```

```python
mae = mean_absolute_error(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))
r2 = r2_score(y, y_pred)

print(f"MAE: {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R²: {r2:.2%}")

# Save the trained model
with open("knn_model.pkl", "wb") as f:
    pickle.dump(model, f)

print("Model saved successfully!")
```
