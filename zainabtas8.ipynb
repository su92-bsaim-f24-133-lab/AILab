```python
import math


def calculate(depth, node, turn_max, numbers, max_depth):

    if depth == max_depth:
        return numbers[node]
    

    if turn_max:
        value1 = calculate(depth + 1, node * 2,     False, numbers, max_depth)
        value2 = calculate(depth + 1, node * 2 + 1, False, numbers, max_depth)
        return max(value1, value2)
    

    else:
        value1 = calculate(depth + 1, node * 2,     True, numbers, max_depth)
        value2 = calculate(depth + 1, node * 2 + 1, True, numbers, max_depth)
        return min(value1, value2)



leaf_nodes = [3, 5, 2, 9]                     
depth_of_tree = int(math.log(len(leaf_nodes), 2))

print("Tree leaves:", leaf_nodes)
best_choice = calculate(0, 0, True, leaf_nodes, depth_of_tree)
print("Optimal value (Max can get):", best_choice)
```
